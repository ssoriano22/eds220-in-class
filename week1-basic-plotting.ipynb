```python
import pandas as pd

# Read in file
df = pd.read_csv('data/wetlands_seasonal_bird_diversity.csv')

# Check the first 5 rows
df.head()
```

## Line Plots

```python
# Default plot(): one line plot per column with numeric data
df.plot()
```

```python
# General Syntax for plotting
# df.plot(x='x_values_column', y='y_values_column')
```

```python
# Bird species registered during winter at CSM yearly
df.plot(x='year', y='CSM_winter')
```

```python
df.plot(x='year', 
        y='CSM_winter',
        title='Bird species registered during winter at Carpinteria Salt Marsh',
        xlabel='Year',
        ylabel='Number of bird species',
        color='green',
        legend=False
        )
```

## Multiple Line Plots

```python
# Compare the bird surveys at the Tijuana Estuary during spring and fall across years
df.plot(x='year', 
        y=['TJE_spring', 'TJE_fall'],
        title = 'Seasonal bird surveys at Tijuana Estuary',
        xlabel='Year',
        ylabel='Number of bird species',        
        color = {'TJE_spring':'#F48FB1',
                 'TJE_fall': '#AB47BC'
                 }
        )
```

```python
# Can be two separate plots too
df.plot(x='year', 
        y=['TJE_spring', 'TJE_fall'],
        title = 'Seasonal bird surveys at Tijuana Estuary',
        xlabel='Year',
        ylabel='Number of bird species',        
        color = {'TJE_spring':'#F48FB1',
                 'TJE_fall': '#AB47BC'
                 },
        subplots=True
        );
```

## Data Exploration

```python
# Read in data
URL = 'https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv'
penguins = pd.read_csv(URL)

penguins.head()
```

```python
# Check column data types and NA values
penguins.info()
```

```python
# Simple statistics about numeric columns
penguins.describe()
```

```python
# Count unique values in categorical columns and year
penguins[['species', 'island', 'sex', 'year']].nunique()
```

```python
# Get unique values in species column
penguins['species'].unique()
```

```python
# Number of values per unique value in species column
penguins['species'].value_counts()
```

## `kind` argument in `plot()`

```python
# `kind` parameter changes the type of plot from the default line
```

## Scatter plots

```python
# Scatterplot: Flipper length mm vs. Body Mass g
penguins.plot(kind='scatter',
              x='flipper_length_mm', 
              y='body_mass_g')
```

```python
# Adding additional customations
penguins.plot(kind='scatter',
              x='flipper_length_mm', 
              y='body_mass_g',
              title='Flipper length and body mass for Palmer penguins',
              xlabel='Flipper length (mm)',
              ylabel='Body mass (g)',
              color='#ff3b01',
              alpha=0.4  # Controls transparency
              )
```

## Bar Plots

```python
# Example of kind = 'bar' w/ smallest mass penguins
smallest = penguins['body_mass_g'].nsmallest(10)
smallest
```

```python
smallest.plot(kind='bar')
```

```python
# Another way to look at 10 smallest mass penguin rows
penguins.nsmallest(10, 'body_mass_g')
```

## Histograms

```python
# Example of kind = 'hist'
# Using plot without subsetting data - a mess again
penguins.plot(kind='hist')
```

```python
# Distribution of flipper length measurements
# First select data, then plot for something that is interpretable
penguins['flipper_length_mm'].plot(kind='hist',
                                title='Penguin flipper lengths',
                                xlabel='Flipper length (mm)',
                                grid=True)
```

## Check-In

1. Select the bill_length_mm and bill_depth_mm columns in the penguins data frame and then update the kind parameter to box to make boxplots of the bill length and bill depth.

2. Create a simple histogram of the flipper length of female gentoo penguins.

```python
# Select bill columns and rename for plotting
bill_df = penguins[['bill_length_mm','bill_depth_mm']]
bill_df = bill_df.rename(columns = {'bill_length_mm':'Bill_Length',
                                    'bill_depth_mm':'Bill_Depth'})
# Plot bill data as boxplot
bill_df.plot(kind = 'box',
             title='Penguin Bill Characteristics',
             ylabel='mm')
```

```python
# Select Gentoo species and flipper_length_mm and rename for plotting
genfflip_df = penguins[['species','flipper_length_mm']][(penguins['species'] == 'Gentoo') &
                                                        (penguins['sex'] == 'female')]
genfflip_df = genfflip_df.rename(columns = {'species':'Species',
                                          'flipper_length_mm':'Flipper Length'})
genfflip_df.plot(kind = 'hist',
                 title='Female Gentoo Penguin Flipper Length Distribution',
                 xlabel = 'Flipper Length (mm)')
```

## Lesson Summary

The basic python plotting method is `.plot()`, which can produce many simple graph types including:
- line (default)
- scatter
- bar
- box
- hist

Calling `.plot()` on a df without subsetting is rarely a good idea because variables are likely on different scales/measured differently.
