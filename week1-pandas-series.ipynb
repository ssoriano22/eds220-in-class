## pandas Series

```python
import pandas as pd
import numpy as np
```

A pandas.Series having an index is the main difference between a pandas.Series and a NumPy array:

```python
# A numpy array
arr = np.random.randn(4) # random values from std normal distribution
print(type(arr))
print(arr, "\n")

# A pandas series made from the previous array
s = pd.Series(arr)
print(type(s))
print(s)
```

```python
# Basic method of create a pandas Series
# s = pd.Series(data, index=index)
```

The data parameter can be:
- a list or NumPy array,
- a Python dictionary, or
- a single number, boolean (True/False), or string.
  
The index parameter is optional. If we wish to include it, it must be a list of indices of the same length as data.

## Creating a pandas Series

```python
# Example: Creating a pandas Series from a np array
pd.Series(np.arange(3), index=[2023, 2024, 2025])
```

```python
# Example: Creating a pandas Series from a python list (default index)
pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
# Example: Creating a pandas Series from a python dictionary
# Make dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}
# Initialize series w/ dictionary
pd.Series(d)
```

```python
# Example: Creating a pandas Series from a single value
pd.Series(3.0, index = ['A', 'B', 'C'])
```

## Simple Operations

```python
# Initialize a pandas Series
s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in the series by 10
print(s /10, '\n')

# Take the exponential of each element in the series
print(np.exp(s), '\n')

# Print original Series
print(s)
```

```python
# Can produce new pandas Series w/ conditionals
s > 70
```

## Identifying Missing Values

In pandas we can represent a missing, NULL, or NA value with the float value numpy.nan, which stands for “not a number”.

```python
# pandas Series with NaNs in it
s = pd.Series([1, 2, np.nan, 4, np.nan])
s
```

```python
# Check if pandas Series has any NaNs using hasnans attribute
s.hasnans
```

```python
# Check pandas Series for NA values, returns Series
s.isna()
```

## Data Frames

It represents tabular data and we can think of it as a spreadsheet. Each column of a pandas.DataFrame is a pandas.Series.

### Creating a pandas DataFrame

```python
# Initialize dictionary with column data (Series) 
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]),
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# Change index of df
df.index = ['a','b','c']
df
```

### Check-In

```python
df.columns = ['C1','C2']
df
```
